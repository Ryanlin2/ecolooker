```python
df["id_missing"] = df["Complaint ID"].isna()
df["id_missing"].mean()   # overall proportion missing
```

```python
def missingness_by_category(df, col, flag="id_missing"):
    out = df.groupby(col, observed=True)[flag].agg(["mean", "size"])
    out = out.rename(columns={"mean": "missing_rate", "size": "n"})
    return out.sort_values("missing_rate", ascending=False)

missingness_by_category(df, "Company response to consumer")
missingness_by_category(df, "Submitted via")
missingness_by_category(df, "Product")
missingness_by_category(df, "State")
```

```python
import pandas as pd


def convert_column_to_datetime(
    df: pd.DataFrame,
    column: str,
) -> None:
    """Convert mixed ISO 8601 strings to UTC datetimes in place.

    Date-only values are interpreted as midnight UTC.
    Invalid non-null values raise ValueError.
    """
    df[column] = pd.to_datetime(
        df[column],
        format="mixed",
        errors="raise",
        utc=True,
    )

    return df
```

```python
df_date = convert_column_to_datetime(df_date, "Date received")
```
