```python
def boolify(df, col):
    mapping = {"Yes": True, "No": False}

    invalid = df.loc[
        df[col].notna() & ~df[col].isin(mapping),
        col
    ].unique()

    if len(invalid) > 0:
        raise ValueError(f"Unexpected values in {col!r}: {invalid.tolist()}")

    df[col] = df[col].map(mapping).astype("boolean")
    return df
```
