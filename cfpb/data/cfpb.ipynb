```python
def missingness_by_category(df, col, flag="id_missing"):
    out = df.groupby(col, observed=True)[flag].agg(["mean", "size"])
    out = out.rename(columns={"mean": "missing_rate", "size": "n"})
    return out.sort_values("missing_rate", ascending=False)

missingness_by_category(df, "Company response to consumer")
missingness_by_category(df, "Submitted via")
missingness_by_category(df, "Product")
missingness_by_category(df, "State")
```

```python
from datetime import datetime

def get_date_format(date_string):
    """
    Detect the format string of a date string.
    Returns the matching strptime format, or None if no known format matches.
    """
    formats = [
        "%Y-%m-%d",           # 2023-05-17
        "%m/%d/%Y",           # 05/17/2023
        "%m/%d/%y",           # 05/17/23
        "%d/%m/%Y",           # 17/05/2023
        "%m-%d-%Y",           # 05-17-2023
        "%d-%m-%Y",           # 17-05-2023
        "%Y/%m/%d",           # 2023/05/17
        "%B %d, %Y",          # May 17, 2023
        "%b %d, %Y",          # May 17, 2023
        "%d %B %Y",           # 17 May 2023
        "%d %b %Y",           # 17 May 2023
        "%Y-%m-%d %H:%M:%S",  # 2023-05-17 14:30:00
        "%Y-%m-%dT%H:%M:%S",  # 2023-05-17T14:30:00 (ISO)
        "%m/%d/%Y %H:%M",     # 05/17/2023 14:30
    ]

    date_string = date_string.strip()
    for fmt in formats:
        try:
            datetime.strptime(date_string, fmt)
            return fmt
        except ValueError:
            continue
    return None
```

```python
def str_column_to_date(df: pd.DataFrame, column: str) -> pd.DataFrame:
    df[column] = pd.to_datetime(
        df[column],
        format="%Y-%m-%d",
        errors="raise"
    ).dt.date
    return df
```
